## Ejercicio 1
El 70% de los clientes ven un anuncio y, de ellos, compra el 30%.
El 30% no ve el anuncio y, de ellos, compra el 10%.
Calcula la probabilidad total de que un cliente compre.

```python
# Definir las probabilidades
p_ver = 0.70
p_compra_dado_ver = 0.30

p_no_ver = 0.30
p_compra_dado_no_ver = 0.10

# Aplicar la fórmula de probabilidad total
p_compra = (p_compra_dado_ver * p_ver) + (p_compra_dado_no_ver * p_no_ver)
print(f"Probabillidad de que un cliente haga una compra = {p_compra:.2f} (o {p_compra * 100:.0f}%)")
```

## Ejercicio 2
 Se sabe que:# El 40% de los días son nublados; de esos, el 75% llueve.
El 60% de los días no son nublados; de esos, el 10% llueve.
 ¿Cuál es la probabilidad total de que llueva?

```python

p_nublado = 0.40
p_llueve_dado_nublado = 0.75

# No nublado
p_no_nublado = 0.60
p_llueve_dado_no_nublado = 0.10

# Probabilidad total de lluvia
p_lluvia = (p_llueve_dado_nublado * p_nublado) + (p_llueve_dado_no_nublado * p_no_nublado)
print(f"Probabilidad de lluvia = {p_lluvia:.2f} (o {p_lluvia * 100:.0f}%)")
```

## Ejercicio 3
Hay dos cajas de audífonos:
Caja 1: 2 sirven y 6 fallan (Total = 8 audífonos) Caja 2: 5 sirven y 3 fallan (Total = 8 audífonos)
Se elige una caja al azar (P(C1) = 0.5, P(C2) = 0.5) y después un audífono.
¿Cuál es la probabilidad de que sirva?

```python


# Probabilidad de elegir cada caja
p_c1 = 0.5
p_c2 = 0.5

# Probabilidad condicional de que sirva dado la caja
p_sirve_dado_c1 = 2 / 8
p_sirve_dado_c2 = 5 / 8

# Probabilidad total de que sirva
p_sirve = (p_sirve_dado_c1 * p_c1) + (p_sirve_dado_c2 * p_c2)
print(f"Pobabilidad de que sirva = {p_sirve:.4f} (o {p_sirve * 100:.2f}%)")
```
