### Ejercicio 1
El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

**Definir de eventos:**
* S: El correo es Spam (P(S) = 0.20)
* D: El correo es Deseado/No Spam (P(D) = 0.80$)
* O: El correo contiene la palabra "oferta"

**Verosimilitudes:**
* P(O|S) = 0.80
* P(O|D) = 0.10

```python

p_s = 0.20  # P(S)
p_d = 0.80  # P(D)

p_o_dado_s = 0.80  # P(O|S)
p_o_dado_d = 0.10  # P(O|D)

# 1. Probabilidad total de recibir la palabra "oferta"
p_o = (p_o_dado_s * p_s) + (p_o_dado_d * p_d)
print(f"Probabilidad total de 'oferta': {p_o:.2f}")

# 2. Aplicamos el Teorema de Bayes: P(S|O)
p_s_dado_o = (p_o_dado_s * p_s) / p_o
print(f"Probabilidad de que sea Spam dado que tiene 'oferta': {p_s_dado_o:.4f} ({p_s_dado_o * 100:.2f}%)")
```

### Ejercicio 2
En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.
Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

**Definición de eventos:**
* C: Navegador Chrome (P(C) = 0.40)
* S: Navegador Safari (P(S) = 0.35)
* O: Otros navegadores (P(O) = 0.25)
* V: Se realiza una Venta

**Tasa de conversión (Verosimilitud de Venta):**
* P(V|C) = 0.03
* P(V|S) = 0.04
* P(V|O) = 0.02

```python
# Definir eventos
# C: Chrome, S: Safari, O: Otros
# V: realiza compra (venta)

p_c = 0.40
p_s = 0.35
p_o = 0.25

p_v_dado_c = 0.03
p_v_dado_s = 0.04
p_v_dado_o = 0.02

#  probabilidad total de venta
p_v = p_v_dado_c * p_c + p_v_dado_s * p_s + p_v_dado_o * p_o
print(f"Probabilidad de venta = {p_v:.4f}")

# plicar Bayes para cada navegador
p_c_dado_v = (p_v_dado_c * p_c) / p_v
p_s_dado_v = (p_v_dado_s * p_s) / p_v
p_o_dado_v = (p_v_dado_o * p_o) / p_v

print(f"probabilidad de Chrome ≈ {p_c_dado_v:.4f}")
print(f"Probabilidad de Safari ≈ {p_s_dado_v:.4f}")
print(f"Proabbilidad de Otros ≈ {p_o_dado_v:.4f}")

# Determinamos el mayor
resultados = {"Chrome": p_c_dado_v, "Safari": p_s_dado_v, "Otros": p_o_dado_v}

```
