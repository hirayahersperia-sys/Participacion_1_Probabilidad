## Ejercicio 1
Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
from itertools import product

# Espacio muestral (dos dados, 6 caras cada uno)
S = list(product(range(1, 7), repeat=2))

# Casos favorables (suma igual a 5)
A = [x for x in S if x[0] + x[1] == 5]


print(f"Total de resultados posibles: {len(S)}")
print(f"Casos favorables (A): {A}")
print(f"Cantidad de casos favorables: {len(A)}")
print(f"Probabilidad P(A): {len(A)} / {len(S)} = {len(A)/len(S):.4f}")
```

## Ejercicio 2
En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.



```python

from itertools import product

# Espacio muestral (contraseñas de 3 dígitos del 0 al 9)
S = list(product(range(10), repeat=3))

# Casos favorables (ningún dígito se repite)
A = [x for x in S if len(set(x)) == 3]


print(f"Total de resultados posibles: {len(S)}")
print(f"Cantidad de casos favorables: {len(A)}")
print(f"Probabilidad P(A): {len(A)} / {len(S)} = {len(A)/len(S):.4f}")
```

## Ejercicio 3
 Tienes 8 canciones distintas y las pones en un orden aleatorio.
Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

```python

from itertools import permutations

# Definimos las 8 canciones (0 es la favorita, 7 es la que menos gusta)
canciones = list(range(8))

# Espacio muestral (todas las permutaciones de las 8 canciones)
S = list(permutations(canciones))

# Casos favorables (canción favorita '0' en primer lugar e indeseada '7' al final)
A = [x for x in S if x[0] == 0 and x[-1] == 7]


print(f"Total de resultados posibles (8!): {len(S)}")
print(f"Cantidad de casos favorables (6!): {len(A)}")
print(f"Probabilidad P(A): {len(A)} / {len(S)} = {len(A)/len(S):.6f}")
```

## Ejercicio 4
 Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o 10?
 ¿Por cuántos resultados le gana una suma a la otra?

```python

from itertools import product

# Espacio muestral de tres dados
S = list(product(range(1, 7), repeat=3))

# Casos favorables para suma 9
A_9 = [x for x in S if sum(x) == 9]
# Casos favorables para suma 10
A_10 = [x for x in S if sum(x) == 10]


print(f"Total de resultados posibles: {len(S)}")
print(f"Casos favorables para suma 9: {len(A_9)}")
print(f"Casos favorables para suma 10: {len(A_10)}")
print(f"Probabilidad de suma 9: {len(A_9)}/{len(S)} = {len(A_9)/len(S):.4f}")
print(f"Probabilidad de suma 10: {len(A_10)}/{len(S)} = {len(A_10)/len(S):.4f}")

diferencia = len(A_10) - len(A_9)
if diferencia > 0:
    print(f"\nEs más probable que la suma sea 10. Le gana por {diferencia} resultados posibles.")
elif diferencia < 0:
    print(f"\nEs más probable que la suma sea 9. Le gana por {abs(diferencia)} resultados posibles.")
else:
    print("\nAmbas sumas tienen la misma probabilidad.")
```

## Ejercicio 5
 Ana, Bruno, Carla y Diego hacen un amigo secreto.
 Calcula la probabilidad de que nadie saque su propio regalo (derangement).

```python

from itertools import permutations

amigos = ["Ana", "Bruno", "Carla", "Diego"]
# Espacio muestral (todas las formas de asignar regalos)
S = list(permutations(amigos))

# Casos favorables: nadie saca su propio nombre
# Si 'amigos[i]' es la persona, 'x[i]' es a quién le regala.
A = [x for x in S if all(x[i] != amigos[i] for i in range(len(amigos)))]


print(f"Total de resultados posibles: {len(S)}")
print(f"Casos favorables (Desajustes): {len(A)}")
print(f"Lista de repartos válidos: {A}")
print(f"Probabilidad P(A): {len(A)} / {len(S)} = {len(A)/len(S):.4f}")
```
