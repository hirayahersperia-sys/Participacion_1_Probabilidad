## Ejercicio 1
Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

```python
import random

def lanzar_dado():
    return random.randint(1, 6)

n1 = 10000
sumas_5 = 0
frecuencias_1 = []

for i in range(1, n1 + 1):
    if lanzar_dado() + lanzar_dado() == 5:
        sumas_5 += 1
    frecuencia = sumas_5 / i
    frecuencias_1.append(frecuencia)

prob_clasica_1 = 4 / 36


print("Número de lanzamientos:", n1)
print(f"Probabilidad empírica de obtener suma 5: {frecuencia:.4f}")
print(f"Probabilidad clásica: {prob_clasica_1:.4f}")
```

## Ejercicio 2
Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

```python
import random

def generar_pin_4():
    return [random.randint(0, 9) for _ in range(4)]

n2 = 10000
sin_repetir = 0
frecuencias_2 = []

for i in range(1, n2 + 1):
    pin = generar_pin_4()
    if len(set(pin)) == 4:
        sin_repetir += 1
    frecuencia = sin_repetir / i
    frecuencias_2.append(frecuencia)

prob_clasica_2 = (10 * 9 * 8 * 7) / (10**4)


print("Número de simulaciones:", n2)
print(f"Probabilidad empírica de PIN sin repetición: {frecuencia:.4f}")
print(f"Probabilidad clásica (fracción): {prob_clasica_2:.4f}")
```

## Ejercicio 3
Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
import random

def lanzar_dado():
    return random.randint(1, 6)

n3 = 20000
suma_9 = 0
suma_10 = 0
frecuencias_suma_9 = []
frecuencias_suma_10 = []

for i in range(1, n3 + 1):
    resultado = lanzar_dado() + lanzar_dado() + lanzar_dado()
    if resultado == 9:
        suma_9 += 1
    elif resultado == 10:
        suma_10 += 1

    freq_9 = suma_9 / i
    freq_10 = suma_10 / i
    frecuencias_suma_9.append(freq_9)
    frecuencias_suma_10.append(freq_10)


print("Número de lanzamientos de tres dados:", n3)
print(f"Frecuencia empírica de suma 9: {freq_9:.4f} ({suma_9} veces)")
print(f"Frecuencia empírica de suma 10: {freq_10:.4f} ({suma_10} veces)")
if freq_10 > freq_9:
    print("Se nota en la simulación que la suma 10 sale más frecuentemente.")
else:
    print("Las frecuencias están muy reñidas. Intente subir los lanzamientos o correr de nuevo.")
```

## Ejercicio 4
Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
import random

urna = ["roja", "roja", "roja", "azul", "azul"]
n4 = 2000
azules_extraidas = 0
frecuencias_4 = []

for i in range(1, n4 + 1):
    esfera = random.choice(urna)
    if esfera == "azul":
        azules_extraidas += 1
    frecuencia = azules_extraidas / i
    frecuencias_4.append(frecuencia)

prob_clasica_4 = 2 / 5


print("Número de extracciones:", n4)
print(f"Probabilidad empírica de obtener una bola azul: {frecuencia:.4f}")
print(f"Probabilidad clásica: {prob_clasica_4:.4f}")
```

## Ejercicio 5
Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
import random

def lanzar_moneda():
    return random.choice(["águila", "sol"])

n5 = 10000
series_con_racha = 0
frecuencias_5 = []

for i in range(1, n5 + 1):
    volados = [lanzar_moneda() for _ in range(20)]
    racha_maxima = 0
    racha_actual = 0
    for v in volados:
        if v == "águila":
            racha_actual += 1
            if racha_actual > racha_maxima:
                racha_maxima = racha_actual
        else:
            racha_actual = 0

    if racha_maxima >= 5:
        series_con_racha += 1

    frecuencia = series_con_racha / i
    frecuencias_5.append(frecuencia)


print("Número de series simuladas:", n5)
print(f"Probabilidad estimada de racha de 5 águilas: {frecuencia:.4f}")
```

## Ejercicio 6
Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
import random

n6 = 500
tiempo_actual = 0
tiempo_salida_anterior = 0
esperas_mas_10 = 0
frecuencias_6 = []

for i in range(1, n6 + 1):
    tiempo_entre_llegadas = random.randint(1, 5)
    tiempo_actual += tiempo_entre_llegadas

    tiempo_atencion = random.randint(2, 4)

    tiempo_inicio_atencion = max(tiempo_actual, tiempo_salida_anterior)
    espera = tiempo_inicio_atencion - tiempo_actual

    if espera > 10:
        esperas_mas_10 += 1

    frecuencia = esperas_mas_10 / i
    frecuencias_6.append(frecuencia)

    tiempo_salida_anterior = tiempo_inicio_atencion + tiempo_atencion

print("Número de clientes:", n6)
print(f"Probabilidad de que un cliente espere más de 10 minutos: {frecuencia:.4f}")
```
