## Ejercicio 1
 Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

```python
espacio = [1, 2, 3, 4, 5, 6]

# B: El número obtenido es par
B = {x for x in espacio if x % 2 == 0}

# A: El número obtenido es 2
A = {x for x in espacio if x == 2}

interseccion = A & B

probabilidad = len(interseccion) / len(B)
print("Ejercicio 1:")
print(f"B (pares) = {B}")
print(f"A (obtener 2) = {A}")
print(f"P(2 | par) = {len(interseccion)} / {len(B)} =", probabilidad)
```

## Ejercicio 2
se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
from itertools import product

# S: Espacio muestral completo de 3 dados
S = list(product(range(1, 7), repeat=3))

# Condición: las tres caras son distintas
distintos = [t for t in S if len(set(t)) == 3]

# Evento: la suma de las caras es 9
suma_9 = [t for t in distintos if sum(t) == 9]

probabilidad = len(suma_9) / len(distintos)
print("Ejercicio 2:")
print(f"Casos con caras distintas: {len(distintos)}")
print(f"Casos que además suman 9: {len(suma_9)}")
print(f"P(suma 9 | caras distintas) = {len(suma_9)} / {len(distintos)} =", probabilidad)
```

## Ejercicio 3
De 100 días: 40 son nublados. De esos, 30 llueve.
¿Cuál es la probabilidad de lluvia dado que está nublado?

```python
dias_totales = 100
dias_nublados = 40
dias_lluvia_y_nublado = 30 # 'De esos (nublados), 30 llueve'

# P(Lluvia | Nublado) = P(Lluvia y Nublado) / P(Nublado)
# Usando frecuencias directas:
probabilidad = dias_lluvia_y_nublado / dias_nublados

print("Ejercicio 3:")
print(f"Días nublados (B): {dias_nublados}")
print(f"Días con lluvia dado que está nublado (A ∩ B): {dias_lluvia_y_nublado}")
print(f"P(lluvia | nublado) = {dias_lluvia_y_nublado} / {dias_nublados} =", probabilidad)
```
